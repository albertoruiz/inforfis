```python
# preparamos la tabla y la modificamos
# en la misma celda para no tener que evaluar
# varias celdas cada vez que hagamos un cambio

tabla = list(range(100))

for k in range(25):
    tabla[3*k] = "X"

print(tabla)
```

```python
# mostramos el progreso de la tabla paso a paso

tabla = list(range(100))

for k in range(25):
    tabla[3*k] = "X"
    print(tabla)
```

```python
# vamos a intentar que los múltiplos lleguen hasta el final

tabla = list(range(100))

for k in range(1, 100/3):
    tabla[3*k] = "X"

print(tabla)
```

```python
# vaya, range solo admite enteros
# tenemos que utilizar la división entera,
# expresada en python con doble barra

tabla = list(range(100))

for k in range(1, 100//3):
    tabla[3*k] = "X"

print(tabla)
```

```python
# pero como el límite superior de range es no inclusivo, sumamos 1

tabla = list(range(100))

for k in range(2, 100//3 + 1):
    tabla[3*k] = "X"

print(tabla)
```

```python
# ya tenemos correctamente tachados los múltiplos de 3
# pero queda muy feo. Es mucho mejor usar el parámetro "step" de range

tabla = list(range(100))

for k in range(3,100,3):
    tabla[k] = "X"

print(tabla)
```

```python
# bien, pero el primer múltiplo debemos dejarlo

tabla = list(range(100))

for k in range(2*5,100,5):
    tabla[k] = "X"

print(tabla)
```

```python
# perfecto, ya solo queda hacer lo mismo con todos los números
# para lo cual usamos otro bucle for. Un bucle de bucles, o bucles anidados

tabla = list(range(100))

for n in range(2,100):
    for k in range(2*n,100,n):
        tabla[k] = "X"

print(tabla)
```

```python
# Problema resuelto. Pero me gustaría limpiar la tabla, quitando las X

tabla = list(range(100))

for n in range(2,100):
    for k in range(2*n,100,n):
        tabla[k] = "X"

#print(tabla)

resultado = [n for n in tabla if n != "X"]
print(resultado)
```

```python
# sin embargo, hay un pequeño detalle de eficiencia que podemos mejorar
# Si tachamos múltiplos de todos los números estamos desperdiciando trabajo

tabla = list(range(26))

for n in range(2,7):
    for k in range(2*n,26,n):
        tabla[k] = "X"
    print(n, tabla)
```

```python
# los múltiplos de 4 y de 6 no cambian la tabla.
# solo hay que tachar múltiplos de números que no estén ya tachados

tabla = list(range(26))

for n in range(2,7):
    if tabla[n] != "X":
        for k in range(2*n,26,n):
            tabla[k] = "X"
        print(n, tabla)
```

```python
# Ya lo tenemos. Pero queda lo más importante: encapsularlo dentro de una función reutilizable,
# que tendrá unos parámetros de entrada bien definidos, y producirá un resultado.
# Internamente trabajará con variables auxiliares (locales), invisibles "desde fuera"

def criba(M):

    tabla = list(range(M))

    for n in range(2,M):
        if tabla[n] != "X":
            for k in range(2*n,M,n):
                tabla[k] = "X"

    return tabla


print(criba(26))
print(criba(200))
```

```python
# añadimos la limpieza de la tabla

def criba(M):

    tabla = list(range(M))

    for n in range(2,M):
        if tabla[n] != "X":
            for k in range(2*n,M,n):
                tabla[k] = "X"

    tabla_limpia = [n for n in tabla if n != "X" and n>1]
    return tabla_limpia

print(criba(26))
print(criba(200))
```

```python
# Estupendo. Ahora podemos usarla para lo que queramos,
# como una instrucción más del lenguaje

def criba(M):

    tabla = list(range(M))

    for n in range(2,M):
        if tabla[n] != "X":
            for k in range(2*n,M,n):
                tabla[k] = "X"

    tabla_limpia = [n for n in tabla if n != "X" and n>1]
    return tabla_limpia


1001 in criba(1001)
```

```python
# por ejemplo, podemos comprobar si un número es primo, viendo si
# se ha quedado en una criba que podría haber llegado hasta él

def criba(M):

    tabla = list(range(M))

    for n in range(2,M):
        if tabla[n] != "X":
            for k in range(2*n,M,n):
                tabla[k] = "X"

    tabla_limpia = [n for n in tabla if n != "X" and n>1]
    return tabla_limpia

def primoQ(n):
    return n in criba(n+1)

primoQ(12), primoQ(13), primoQ(997)
```

```python
# Los bucles pueden hacerse con instrucciones más primitivas

N = 100
tabla = list(range(N+1))

k = 2
position = k*2
while position <= N:
    tabla[position] = 'X'
    position = position + k

print(tabla)
```

```python
# pero normalmente los bucles for (cuando son apropiados),
# producen un código más claro

N = 100
tabla = list(range(N+1))

n = 2
while n < N:
    k = n
    position = k*2
    while position <= N:
        tabla[position] = 'X'
        position = position + k
    n = n + 1

print(tabla)
```
