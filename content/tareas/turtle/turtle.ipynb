```python
# Necesario en jupyterlite
import sys, os
sys.path.insert(0, os.getcwd())

from jupyturtle import *

def save(filename):
    if filename[-4:] != '.svg':
        filename = filename + ".svg"
    with open(filename,"w") as file:
        file.write(get_turtle().get_SVG().replace('<svg', '<svg xmlns="http://www.w3.org/2000/svg"'))
```

```python
%%turtle 300

for k in range(3):
    forward(100)
    left(90)
```

```python
save("prueba")
```
